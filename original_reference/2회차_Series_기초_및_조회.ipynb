```python
# 1. Series 생성과 인덱스 지정
```

```python
import pandas as pd
s = pd.Series([100, 200, 300], index=['국어', '수학', '영어'])
print(s)
```

```python
# 2. 특정 인덱스 조회
```

```python
print("수학 점수:", s['수학'])
```

```python
# 3. 슬라이싱 #리스트 슬라이싱하는 것과 다른 점 . 뒷부분이 영어 하나 앞이 아니라..뒷부분까지임!
```

```python
print(s['국어':'영어'])
```

```python
# 4. 조건 조회
```

```python
print('200점 이상 과목:\n', s[s >= 200])


#벡터연산 중 Boolean Indexing(불리언 인덱싱) 또는 필터링이라고 함.
```

```python
# 📝 연습: '과학', '사회' 포함한 새로운 Series 만들어보고 250점 이상만 출력해보세요.
```

```python
s2 = pd.Series([300, 200, 280, 190, 230],index=['과학', '사회', '국어', '수학','영어'])
print("250점 이상 과목:\n", s2[s2>=250])
```
