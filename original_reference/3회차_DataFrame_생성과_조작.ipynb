```python
# 1. DataFrame 생성
```

```python
import pandas as pd
data = {'이름': ['철수', '영희', '민수'], '국어': [80, 90, 70], '수학': [75, 85, 95]}
df = pd.DataFrame(data)
print(df)
```

```python
# 2. 열 추가
```

```python
df['영어'] = [88, 92, 85]
print(df)
```

```python
# 3. 열 삭제
```

```python
df = df.drop(columns='국어')
print(df)
```

```python
# 4. 정렬
```

```python
print(df.sort_values(by='수학', ascending=False))
```
