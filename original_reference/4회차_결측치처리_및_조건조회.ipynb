```python
# 1. 결측치 포함한 DataFrame 만들기
```

```python
import pandas as pd
df = pd.DataFrame({'이름': ['철수', '영희', '민수'], '점수': [90, None, 80]})
print(df)
```

```python
# 2. 결측치 확인
```

```python
print(df.isnull())
```

```python
# 3. 결측치 채우기
```

```python
df['점수'] = df['점수'].fillna(0)
print(df)
```

```python
# 4. 조건 조회 (점수 >= 85)
```

```python
print(df[df['점수'] >= 85])
```
