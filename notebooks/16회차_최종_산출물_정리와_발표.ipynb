```python
# 공통 준비: 라이브러리와 데이터 경로
from pathlib import Path
import numpy as np
import pandas as pd
DATA = Path('../data')
print('데이터 폴더:', DATA.resolve())
```

```python
import matplotlib.pyplot as plt
df=pd.read_csv(DATA/'project_activity.csv')
summary=df.groupby('분야',as_index=False).agg(학생수=('학생','count'),평균출석=('출석률','mean'),평균발표=('발표점수','mean')).round(1)
display(summary)
summary.to_csv(DATA/'final_summary.csv',index=False,encoding='utf-8-sig')
ax=summary.plot.bar(x='분야',y='평균발표',legend=False,color='#65a765',ylim=(0,100),title='분야별 평균 발표점수')
ax.set_ylabel('점수'); plt.tight_layout(); plt.savefig(DATA/'final_chart.png',dpi=150,bbox_inches='tight')
```
