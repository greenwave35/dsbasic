```python
# 공통 준비: 라이브러리와 데이터 경로
from pathlib import Path
import numpy as np
import pandas as pd
DATA = Path('../data')
print('데이터 폴더:', DATA.resolve())
```

```python
scores=pd.Series({'국어':82,'영어':91,'수학':76,'과학':95,'사회':88})
print('가산점 후\n',scores+3)
display(scores[scores>=85])
```

```python
# 실습: 평균 이상 과목만 내림차순으로 출력
mean=scores.mean(); selected=scores[scores>=mean].sort_values(ascending=False)
print('평균',mean); display(selected)
```
