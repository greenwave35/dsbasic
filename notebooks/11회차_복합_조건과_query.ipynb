# 11회차 | 복합 조건과 query

**학습 목표**
- and/or 조건을 &, |로 결합한다.
- query로 읽기 쉬운 조건식을 작성한다.

> 실행 순서: 위에서 아래로 `Shift+Enter`를 누르세요. 오류가 나면 먼저 현재 폴더가 `notebooks`인지 확인하세요. 데이터 경로는 `../data`입니다.

```python
# 공통 준비: 라이브러리와 데이터 경로
from pathlib import Path
import numpy as np
import pandas as pd
DATA = Path('../data')
print('데이터 폴더:', DATA.resolve())
```

```python
df=pd.read_csv(DATA/'student_scores.csv')
selected=df[(df['영어']>=90)&(df['과학']>=80)]
display(selected)
display(df.query('국어 >= 90 and 사회 >= 85'))
```

```python
# 실습: 수학이 결측이 아니고 영어 또는 과학이 95 이상
cond=df['수학'].notna() & ((df['영어']>=95)|(df['과학']>=95))
display(df.loc[cond,['이름','영어','수학','과학']])
```

## 90분 상세 수업 운영

### ① 문장을 조건식으로 — 10분
복합 문장을 작은 조건으로 나누고 괄호 구조를 종이에 먼저 쓴다.

### ② 조건 중간표 — 15분
각 조건의 True/False와 최종 조건을 하나의 DataFrame으로 확인한다. 틀린 결과가 어느 조건에서 시작됐는지 찾는다.

### ③ 조건 도구 — 20분
`between`, `isin`, `notna`, `str.startswith`를 사용해 범위·목록·결측·문자열 조건을 만든다.

### ④ query 비교 — 15분
일반 불리언 조건과 `query` 결과가 같은지 확인한다. 열 이름에 공백이 있을 때의 주의점을 설명한다.

### ⑤ 모든/하나라도 — 20분
점수 열 전체에 `all(axis=1)`, `any(axis=1)`을 적용해 모든 과목 80 이상, 한 과목이라도 95 이상인 학생을 찾는다.

### ⑥ 오류 찾기 — 10분
`and/or` 사용, 괄호 누락, 결측 비교의 오류를 수정한다.
