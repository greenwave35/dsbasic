```python
# 공통 준비: 라이브러리와 데이터 경로
from pathlib import Path
import numpy as np
import pandas as pd
DATA = Path('../data')
print('데이터 폴더:', DATA.resolve())
```

```python
df=pd.DataFrame({'이름':['철수','영희','민수'],'국어':[80,90,70],'수학':[75,85,95]})
df['영어']=[88,92,85]; df['평균']=df[['국어','수학','영어']].mean(axis=1)
display(df)
display(df.drop(columns='국어'))
```

```python
# 실습: 평균 열을 맨 앞으로 이동
cols=['이름','평균','국어','수학','영어']; display(df[cols])
```
