**학습 목표**
- 두 수치 변수의 관계를 산점도로 본다.
- 색·크기·주석으로 정보를 추가한다.

> 실행 순서: 위에서 아래로 `Shift+Enter`를 누르세요. 오류가 나면 먼저 현재 폴더가 `notebooks`인지 확인하세요. 데이터 경로는 `../data`입니다.

```python
# 공통 준비: 라이브러리와 데이터 경로
from pathlib import Path
import numpy as np
import pandas as pd
DATA = Path('../data')
print('데이터 폴더:', DATA.resolve())
```

```python
import matplotlib.pyplot as plt
df=pd.read_csv(DATA/'study_time.csv')
plt.figure(figsize=(7,5))
plt.scatter(df['공부시간'],df['점수'],s=df['수면시간']*18,c=df['점수'],cmap='viridis',edgecolor='black')
for _,r in df.iterrows(): plt.annotate(r['학생'],(r['공부시간'],r['점수']),xytext=(4,4),textcoords='offset points')
plt.xlabel('공부 시간'); plt.ylabel('점수'); plt.title('공부 시간과 점수'); plt.colorbar(label='점수'); plt.grid(alpha=.25);
```

```python
# 실습: 상관계수 확인(인과관계와는 다름)
print(df[['공부시간','수면시간','점수']].corr(numeric_only=True).round(2))
```

## 90분 상세 수업 운영

### ① 점 하나의 의미 — 10분
산점도의 한 점이 학생 한 명의 두 변수 값임을 설명한다. 점을 선으로 잇지 않는 이유를 말한다.

### ② 기본 산점도 — 15분
공부시간과 점수를 x·y축에 두고 제목·축·격자를 설정한다. 관계의 방향과 강도를 말로 표현한다.

### ③ 상관계수 — 15분
1, -1, 0의 뜻을 배우고 공부시간·수면시간·점수 상관표를 읽는다. 상관과 인과를 구분한다.

### ④ 정보 추가 — 15분
점 크기는 수면시간, 색은 점수, 주석은 학생 이름으로 표현한다. 정보가 너무 많아질 때 가독성을 평가한다.

### ⑤ 이상치 실험 — 20분
가상 이상치 한 행을 추가하고 상관계수 변화를 비교한다. 작은 표본에서 한 점의 영향이 큰 이유를 설명한다.

### ⑥ 종합 미션 — 15분
수면시간과 점수 산점도를 만들고 관찰 사실 2문장, 단정할 수 없는 점 1문장을 작성한다.
