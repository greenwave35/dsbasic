```python
# 공통 준비: 라이브러리와 데이터 경로
from pathlib import Path
import numpy as np
import pandas as pd
DATA = Path('../data')
print('데이터 폴더:', DATA.resolve())
```

```python
import matplotlib.pyplot as plt
df=pd.read_csv(DATA/'monthly_activity.csv')
plt.rcParams['axes.unicode_minus']=False
fig,ax=plt.subplots(1,2,figsize=(11,4))
ax[0].bar(df['월'],df['게시글수'],color='#4f86c6'); ax[0].set_title('월별 게시글 수'); ax[0].tick_params(axis='x',rotation=45)
ax[1].plot(df['월'],df['접속수'],marker='o'); ax[1].set_title('월별 접속 추이'); ax[1].tick_params(axis='x',rotation=45)
plt.tight_layout()
```

```python
survey=pd.read_csv(DATA/'survey.csv'); counts=survey['관심분야'].value_counts()
counts.plot.pie(autopct='%1.0f%%',ylabel='',title='관심 분야 비율');
```

## 90분 상세 수업 운영

### ① 질문에 맞는 그래프 — 10분

### ② 막대그래프 — 15분
월별 게시글 수를 그리고 제목·축 이름·단위·격자를 넣는다. y축 0 시작 여부를 점검한다.

### ③ 선그래프 — 15분
