## 프로젝트 질문
**출석률과 과제 완료 수가 발표 점수와 어떤 관계가 있는가?**

예측을 먼저 적고, 아래 코드를 실행한 뒤 예측과 결과를 비교하세요.

```python
import matplotlib.pyplot as plt
df=pd.read_csv(DATA/'project_activity.csv')
display(df.select_dtypes(include='number').describe().round(1))
print(df[['출석률','과제완료','발표점수']].corr().round(2))
plt.scatter(df['출석률'],df['발표점수'],s=df['과제완료']*25,alpha=.75)
for _,r in df.iterrows(): plt.annotate(r['학생'],(r['출석률'],r['발표점수']))
plt.xlabel('출석률'); plt.ylabel('발표점수'); plt.title('연구회 활동 분석'); plt.grid(alpha=.2);
```
