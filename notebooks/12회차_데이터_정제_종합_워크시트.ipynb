```python
# 공통 준비: 라이브러리와 데이터 경로
from pathlib import Path
import numpy as np
import pandas as pd
DATA = Path('../data')
print('데이터 폴더:', DATA.resolve())
```

```python
df=pd.read_csv(DATA/'student_scores.csv')
score_cols=['국어','영어','수학','과학','사회']
df[score_cols]=df[score_cols].fillna(df[score_cols].median())
df['평균']=df[score_cols].mean(axis=1).round(1)
result=df.loc[df['평균']>=85].sort_values('평균',ascending=False)
display(result)
result.to_csv(DATA/'cleaned_high_scores.csv',index=False,encoding='utf-8-sig')
```
