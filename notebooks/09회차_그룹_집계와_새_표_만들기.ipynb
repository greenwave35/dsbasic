# 9회차 | 그룹 집계와 새 표 만들기

**학습 목표**
- groupby로 범주별 통계를 낸다.
- agg 결과를 해석하고 저장한다.

> 실행 순서: 위에서 아래로 `Shift+Enter`를 누르세요. 오류가 나면 먼저 현재 폴더가 `notebooks`인지 확인하세요. 데이터 경로는 `../data`입니다.

```python
# 공통 준비: 라이브러리와 데이터 경로
from pathlib import Path
import numpy as np
import pandas as pd
DATA = Path('../data')
print('데이터 폴더:', DATA.resolve())
```

```python
df=pd.read_csv(DATA/'project_activity.csv')
display(df)
summary=df.groupby('분야').agg(학생수=('학생','count'),평균출석률=('출석률','mean'),평균발표=('발표점수','mean')).round(1)
display(summary)
summary.to_csv(DATA/'field_summary.csv',encoding='utf-8-sig')
```

```python
# 실습: 분야별 과제완료의 합계와 최대값
display(df.groupby('분야')['과제완료'].agg(['sum','max']))
```

## 90분 상세 수업 운영

### ① 전체 평균과 그룹 평균 — 10분
두 평균이 답하는 질문이 다름을 확인하고 그룹 기준을 먼저 정한다.

### ② groupby 기본 — 15분
분야별 학생수·출석률·발표점수를 각각 집계한다. 그룹이 결과표 한 행이 되는 구조를 설명한다.

### ③ 여러 집계 — 20분
`agg`로 평균·최솟값·최댓값·합계를 한 번에 계산한다. 이름 있는 집계로 읽기 좋은 열 이름을 만든다.

### ④ transform — 15분
분야 평균을 원래 각 학생 행에 붙이고 분야 평균 이상 여부를 만든다. `agg`와 결과 행 수를 비교한다.

### ⑤ 종합 미션 — 20분
출석률 85 이상만 대상으로 분야별 학생수·평균 발표·최고 발표를 구하고 CSV로 저장한다.

### ⑥ 해석 — 10분
