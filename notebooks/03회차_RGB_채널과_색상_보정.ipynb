- 채널별 보정과 마스킹을 수행한다.

```python
# 공통 준비: 라이브러리와 데이터 경로
from pathlib import Path
import numpy as np
import pandas as pd
DATA = Path('../data')
print('데이터 폴더:', DATA.resolve())
```

```python
import matplotlib.pyplot as plt
y,x=np.mgrid[0:120,0:180]
rgb=np.zeros((120,180,3),dtype=float)
rgb[...,0]=x/179; rgb[...,1]=y/119; rgb[...,2]=0.35
gain=np.array([1.0,0.75,1.25])
corrected=np.clip(rgb*gain,0,1)
fig,ax=plt.subplots(1,2,figsize=(8,3)); ax[0].imshow(rgb); ax[1].imshow(corrected)
ax[0].set_title('원본'); ax[1].set_title('채널 보정')
for a in ax:a.axis('off')
```

```python
# 실습: 빨강 채널이 0.6보다 큰 픽셀을 노란색으로 표시
marked=corrected.copy(); mask=marked[...,0]>0.6; marked[mask]=[1,1,0]
plt.imshow(marked); plt.axis('off');
```

## 90분 상세 수업 운영

### ① RGB 색 맞히기 — 10분
`[255,0,0]`, `[0,255,0]`, `[0,0,255]`, `[255,255,255]`의 색을 맞힌다. 한 픽셀이 숫자 3개라는 점을 확인한다.

### ② RGB 배열 구조 — 15분
`(높이, 너비, 3)` shape과 마지막 채널축을 설명한다. `rgb[...,0]`으로 빨강 채널만 꺼내어 회색조로 관찰한다.

### ③ 채널별 브로드캐스팅 — 15분
`gain=[1.0,0.7,1.3]`을 곱하고 각 채널 평균 변화를 확인한다. 1회차의 `(3,)` 브로드캐스팅과 연결한다.

### ④ 색 조건 마스크 — 20분
빨강이 150보다 크고 초록보다 30 이상 큰 픽셀을 선택한다. 마스크 shape과 True 개수를 확인하고 선택 영역을 노란색으로 바꾼다.

### ⑤ 종합 실습 — 20분
파랑 강조, 초록 영역 탐색, 선택 영역 색 변경을 연결한다. 빠른 학생은 조건을 두 개 이상 결합해 자신만의 필터를 만든다.

### ⑥ 정리 — 10분
그레이스케일과 RGB shape 차이, 채널 gain이 전체 픽셀에 적용되는 이유, 마스크의 역할을 답한다.
