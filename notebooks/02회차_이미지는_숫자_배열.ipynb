```python
# 공통 준비: 라이브러리와 데이터 경로
from pathlib import Path
import numpy as np
import pandas as pd
DATA = Path('../data')
print('데이터 폴더:', DATA.resolve())
```

```python
import matplotlib.pyplot as plt
img=plt.imread('../images/gradient.pgm').astype(float)/255
print(img.shape, img.min(), img.max())
fig,ax=plt.subplots(1,3,figsize=(9,3))
ax[0].imshow(img,cmap='gray',vmin=0,vmax=1); ax[0].set_title('원본')
ax[1].imshow(np.clip(img+0.25,0,1),cmap='gray'); ax[1].set_title('밝게')
ax[2].imshow(1-img,cmap='gray'); ax[2].set_title('반전')
for a in ax:a.axis('off')
plt.tight_layout()
```

```python
# 실습: 임계값보다 밝은 픽셀만 흰색으로 만드는 이진화
threshold=0.55
binary=(img>=threshold).astype(float)
plt.imshow(binary,cmap='gray',vmin=0,vmax=1); plt.axis('off');
```
