# 8회차 | 행·열 선택과 정렬

**학습 목표**
- loc와 iloc로 행과 열을 선택한다.
- 여러 기준으로 정렬한다.

> 실행 순서: 위에서 아래로 `Shift+Enter`를 누르세요. 오류가 나면 먼저 현재 폴더가 `notebooks`인지 확인하세요. 데이터 경로는 `../data`입니다.

```python
# 공통 준비: 라이브러리와 데이터 경로
from pathlib import Path
import numpy as np
import pandas as pd
DATA = Path('../data')
print('데이터 폴더:', DATA.resolve())
```

```python
df=pd.read_csv(DATA/'student_scores.csv')
display(df)
display(df.loc[1:3,['이름','수학','과학']])
display(df.iloc[:3,:3])
display(df.sort_values(['영어','국어'],ascending=[False,False]))
```

```python
# 실습: 수학 상위 3명의 이름·수학 점수
display(df.nlargest(3,'수학')[['이름','수학']])
```
