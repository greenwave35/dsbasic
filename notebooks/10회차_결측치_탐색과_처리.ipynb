```python
# 공통 준비: 라이브러리와 데이터 경로
from pathlib import Path
import numpy as np
import pandas as pd
DATA = Path('../data')
print('데이터 폴더:', DATA.resolve())
```

```python
df=pd.read_csv(DATA/'student_scores.csv')
display(df); print(df.isna().sum())
filled=df.copy(); filled['수학']=filled['수학'].fillna(filled['수학'].median())
display(filled)
```

```python
# 실습: 각 점수 열의 결측치를 해당 열 평균으로 채우기
score_cols=['국어','영어','수학','과학','사회']
practice=df.copy(); practice[score_cols]=practice[score_cols].fillna(practice[score_cols].mean())
display(practice)
```
