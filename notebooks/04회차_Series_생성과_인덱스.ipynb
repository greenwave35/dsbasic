**학습 목표**
- Series를 생성하고 사용자 인덱스를 지정한다.
- label 조회와 위치 조회를 구분한다.

> 실행 순서: 위에서 아래로 `Shift+Enter`를 누르세요. 오류가 나면 먼저 현재 폴더가 `notebooks`인지 확인하세요. 데이터 경로는 `../data`입니다.

```python
# 공통 준비: 라이브러리와 데이터 경로
from pathlib import Path
import numpy as np
import pandas as pd
DATA = Path('../data')
print('데이터 폴더:', DATA.resolve())
```

```python
s=pd.Series([100,200,300],index=['국어','수학','영어'],name='점수')
display(s)
print('label:',s.loc['수학'],' position:',s.iloc[1])
```

```python
# 실습: 과학·사회를 포함한 Series를 만들고 앞의 3개 값을 조회
s2=pd.Series([300,200,280,190,230],index=['과학','사회','국어','수학','영어'])
display(s2.iloc[:3])
```

`loc`는 라벨, `iloc`는 위치라는 원칙을 단일 조회와 슬라이싱에 적용한다. 라벨 슬라이스는 끝 포함, 위치 슬라이스는 끝 제외임을 비교한다.

### ④ 없는 인덱스 처리 — 10분
