**학습 목표**
- CSV를 읽고 한 열을 Series로 추출한다.
- 기술통계와 value_counts를 사용한다.

> 실행 순서: 위에서 아래로 `Shift+Enter`를 누르세요. 오류가 나면 먼저 현재 폴더가 `notebooks`인지 확인하세요. 데이터 경로는 `../data`입니다.

```python
# 공통 준비: 라이브러리와 데이터 경로
from pathlib import Path
import numpy as np
import pandas as pd
DATA = Path('../data')
print('데이터 폴더:', DATA.resolve())
```

```python
df=pd.read_csv(DATA/'survey.csv')
display(df)
s=df['만족도']; print(s.describe())
display(df['관심분야'].value_counts())
```

```python
# 실습: 만족도가 4 이상인 응답자의 관심분야 빈도
high=df.loc[df['만족도']>=4,'관심분야']
display(high.value_counts())
```
